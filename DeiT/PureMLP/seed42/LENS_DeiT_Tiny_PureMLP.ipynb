# LENS / DeiT-Tiny: Joint LearnedNorm + Direct Pure-MLP Attention

This notebook starts from the official pretrained **timm DeiT-Tiny**
(`deit_tiny_patch16_224.fb_in1k`) and converts it directly to the current
full LENS design.

The conversion is kept consistent with the CIFAR-10 / CIFAR-100 experiments:

1. replace every LayerNorm with LearnedNorm;
2. replace analytical attention Softmax / exponential mapping with the direct pure-MLP score transformation;
3. **unfreeze the whole model immediately**;
4. jointly fine-tune all parameters end-to-end with one learning rate.

There is **no three-stage training** and no frozen-backbone warm-up.

LearnedNorm uses

\[
[a_t,b_t] = f_\theta(x_t), \qquad y_t = a_t x_t + b_t.
\]

The slope `a` keeps the `[0.05, 12]` clamp. The dynamic bias `b`
is **not clamped**.

Direct pure-MLP attention uses the complete score row to predict one
row state, then combines that state with each individual score through
a second shared MLP:

\[
s_q = F_\theta(s_{q,:}),
\qquad
u_{qi} = G_\phi(s_{qi}, s_q),
\qquad
A_{qi} = \frac{\sigma(u_{qi})}{2^{\lfloor\log_2 N\rfloor}}.
\]

For DeiT-Tiny, `N=197`, so the fixed divisor is `128` and every
attention row satisfies

\[
0 < \sum_i A_{qi} < 197/128 = 1.5390625.
\]

There is no Softmax, `exp`, or `exp2` in this score-to-weight mapping.
All hidden activations in both MLPs are ReLU; Sigmoid is used only on
the final scalar output of the second MLP.

Runtime additions:

- every completed epoch is appended to `training_log.txt`;
- `last.pt` is saved every epoch and training auto-resumes from it;
- checkpoints are loaded through CPU first, so a run trained on `cuda:0`
  can be resumed/evaluated on `cuda:1` or another visible device;
- checkpoint metadata is checked to prevent loading the wrong LENS variant.

## 3. Learned replacements

### LearnedNorm

\[
[a_t,b_t] = f_\theta(x_t), \qquad y_t=a_t x_t+b_t.
\]

`a_t` is clamped to `[0.05, 12]`. `b_t` is **unclamped**.

### Direct pure-MLP attention

For every complete attention-score row:

\[
s_{q,:}\in\mathbb{R}^{197}
\rightarrow 197\rightarrow32\rightarrow1
\rightarrow s_q.
\]

Then each score is transformed by the same tiny element MLP:

\[
[s_{qi},s_q]
\rightarrow 2\rightarrow16\rightarrow1
\rightarrow u_{qi},
\qquad
A_{qi}=\frac{\sigma(u_{qi})}{128}.
\]

Both hidden layers use ReLU. Sigmoid is used only after the final output
of the second MLP. There is no analytical row normalization and no
`exp` / `exp2`.

Because `N=197` and the divisor is `128`,

\[
0 < A_{qi} < 1/128,
\qquad
0 < \sum_i A_{qi} < 197/128 = 1.5390625.
\]


```python
# ============================================================
# Learned LayerNorm replacement
# ============================================================

class LearnedLayerNormReplacement(nn.Module):
    """
    Dynamic affine LayerNorm replacement:
        [a_t, b_t] = MLP(x_t)
        y_t = a_t * x_t + b_t

    a_t is clamped; b_t is intentionally not clamped.
    """

    def __init__(self, dim, hidden_dim=32, name="ln"):
        super().__init__()
        self.dim = dim
        self.name = name

        self.predictor_0 = nn.Linear(dim, hidden_dim, bias=True)
        self.predictor_relu = nn.ReLU()
        self.predictor_1 = nn.Linear(hidden_dim, 2, bias=True)

        nn.init.kaiming_uniform_(
            self.predictor_0.weight,
            nonlinearity="relu",
        )
        nn.init.zeros_(self.predictor_0.bias)

        nn.init.normal_(
            self.predictor_1.weight,
            mean=0.0,
            std=1e-3,
        )

        with torch.no_grad():
            self.predictor_1.bias[0] = 1.0
            self.predictor_1.bias[1] = 0.0

    def forward(self, x):
        state = self.predictor_0(x)
        state = self.predictor_relu(state)
        state = self.predictor_1(state)

        raw_a = state[..., 0:1]
        b = state[..., 1:2]  # no beta/b clamp

        a = torch.clamp(
            raw_a,
            min=LEARNED_NORM_A_MIN,
            max=LEARNED_NORM_A_MAX,
        )

        return a * x + b


# ============================================================
# Direct pure-MLP attention replacement, multi-head version
# ============================================================

class DirectMLPAttentionReplacement(nn.Module):
    """
    scores: [B, H, N_query, N_key]

    For every complete score row Z = [z_1, ..., z_N]:

        s = RowMLP(Z)
        u_i = ElementMLP([z_i, s])
        A_i = sigmoid(u_i) / C_N

    where

        C_N = 2 ** floor(log2(N)).

    The row MLP learns one scalar context state from the complete score
    row. The second shared MLP learns the score-to-weight transformation
    conditioned on that row state. No Softmax, exp, or exp2 is used.
    Sigmoid is applied only to the final scalar output to guarantee
    positive bounded coefficients and a bounded aggregate row gain.
    """

    def __init__(
        self,
        num_tokens,
        row_hidden_dim=32,
        element_hidden_dim=16,
    ):
        super().__init__()
        self.num_tokens = int(num_tokens)
        self.gain_divisor = float(
            2 ** math.floor(math.log2(self.num_tokens))
        )

        # Whole-row context/state predictor: N -> hidden -> 1.
        self.row_state_mlp = nn.Sequential(
            nn.Linear(self.num_tokens, row_hidden_dim),
            nn.ReLU(),
            nn.Linear(row_hidden_dim, 1),
        )

        # Shared per-element mapping: [z_i, row_state] -> hidden -> 1.
        # Sigmoid is intentionally applied only in forward().
        self.element_mlp = nn.Sequential(
            nn.Linear(2, element_hidden_dim),
            nn.ReLU(),
            nn.Linear(element_hidden_dim, 1),
        )

        # Small zero-centered initialization keeps the initial final logits
        # near zero, so sigmoid starts near 0.5 while gradients can flow
        # through both MLPs immediately.
        for module in self.modules():
            if isinstance(module, nn.Linear):
                nn.init.trunc_normal_(module.weight, std=0.02)
                if module.bias is not None:
                    nn.init.zeros_(module.bias)

    def forward(self, scores):
        if scores.ndim != 4:
            raise ValueError(
                f"Expected scores [B,H,Nq,Nk], got {tuple(scores.shape)}"
            )

        B, H, Nq, Nk = scores.shape

        if Nk != self.num_tokens:
            raise ValueError(
                f"DirectMLPAttentionReplacement expects N_key={self.num_tokens}, "
                f"but received N_key={Nk}."
            )

        # 1) Predict one scalar context state from each complete row.
        rows = scores.reshape(B * H * Nq, Nk)
        row_state = self.row_state_mlp(rows)
        row_state = row_state.reshape(B, H, Nq, 1)

        # 2) Transform every score conditioned on the shared row state.
        # All elements are processed in parallel by one shared tiny MLP.
        expanded_state = row_state.expand(-1, -1, -1, Nk)
        element_input = torch.stack(
            [scores, expanded_state],
            dim=-1,
        )
        logits = self.element_mlp(element_input).squeeze(-1)

        # 3) Bound the final coefficient only; no analytical row
        # normalization is performed.
        attn = torch.sigmoid(logits) / self.gain_divisor

        return attn
```

## 4. timm attention wrapper with direct pure-MLP attention

The wrapper explicitly reconstructs Q/K/V attention so the analytical
Softmax / exponential score mapping can be replaced. It reuses the pretrained
QKV, projection, normalization, dropout, and optional gate modules by reference.

Each attention block gets its own `197 -> 32 -> 1` row-state predictor and
its own shared `2 -> 16 -> 1` element predictor. Both are shared across that
block's heads and query rows.


```python
# ============================================================
# Wrapper around timm Attention that preserves pretrained weights
# ============================================================

class TimmAttentionWithDirectMLP(nn.Module):
    """
    Drop-in wrapper for timm ViT/DeiT Attention.
    Only the analytical Softmax / exponential score mapping is replaced.
    """

    def __init__(
        self,
        original_attn,
        num_tokens,
        row_hidden_dim=32,
        element_hidden_dim=16,
    ):
        super().__init__()

        self.num_heads = original_attn.num_heads
        self.head_dim = original_attn.head_dim
        self.attn_dim = getattr(
            original_attn,
            "attn_dim",
            self.num_heads * self.head_dim,
        )
        self.scale = original_attn.scale

        self.qkv = original_attn.qkv
        self.q_norm = original_attn.q_norm
        self.k_norm = original_attn.k_norm
        self.attn_drop = original_attn.attn_drop
        self.norm = getattr(original_attn, "norm", nn.Identity())
        self.gate = getattr(original_attn, "gate", None)
        self.proj = original_attn.proj
        self.proj_drop = original_attn.proj_drop

        self.learned_attention = DirectMLPAttentionReplacement(
            num_tokens=num_tokens,
            row_hidden_dim=row_hidden_dim,
            element_hidden_dim=element_hidden_dim,
        )

    def forward(
        self,
        x,
        attn_mask=None,
        is_causal=False,
    ):
        B, N, C = x.shape

        gate = self.gate(x).sigmoid() if self.gate is not None else None

        qkv = self.qkv(x).reshape(
            B,
            N,
            3,
            self.num_heads,
            self.head_dim,
        ).permute(2, 0, 3, 1, 4)

        q, k, v = qkv.unbind(0)
        q = self.q_norm(q)
        k = self.k_norm(k)

        q = q * self.scale
        scores = q @ k.transpose(-2, -1)

        attn_bias = resolve_self_attn_mask(
            N,
            scores,
            attn_mask,
            is_causal,
        )
        scores = maybe_add_mask(scores, attn_bias)

        attn = self.learned_attention(scores)
        attn = self.attn_drop(attn)

        x = attn @ v
        x = x.transpose(1, 2).reshape(B, N, self.attn_dim)

        x = self.norm(x)
        if gate is not None:
            x = x * gate

        x = self.proj(x)
        x = self.proj_drop(x)
        return x
```

## 5. Automatic replacement helpers

LayerNorm replacement walks the whole model recursively. Attention replacement
wraps each DeiT attention block while preserving its pretrained QKV/projection
parameters. After both replacements, the **entire model is trainable**.


```python
# ============================================================
# Automatic replacement helpers
# ============================================================

def _is_layernorm(module):
    return isinstance(module, (nn.LayerNorm, TimmLayerNorm))


def replace_all_layernorms(
    module,
    hidden_dim=32,
    prefix="",
    replaced=None,
):
    if replaced is None:
        replaced = []

    for child_name, child in list(module.named_children()):
        full_name = f"{prefix}.{child_name}" if prefix else child_name

        if isinstance(child, LearnedLayerNormReplacement):
            continue

        if _is_layernorm(child):
            normalized_shape = child.normalized_shape

            if isinstance(normalized_shape, int):
                dim = normalized_shape
            else:
                if len(normalized_shape) != 1:
                    raise ValueError(
                        f"Only 1-D LayerNorm is supported, "
                        f"got {normalized_shape} at {full_name}"
                    )
                dim = normalized_shape[0]

            setattr(
                module,
                child_name,
                LearnedLayerNormReplacement(
                    dim=dim,
                    hidden_dim=hidden_dim,
                    name=full_name,
                ),
            )
            replaced.append(full_name)
        else:
            replace_all_layernorms(
                child,
                hidden_dim=hidden_dim,
                prefix=full_name,
                replaced=replaced,
            )

    return replaced


def replace_all_attention_mapping(
    model,
    row_hidden_dim=32,
    element_hidden_dim=16,
):
    num_tokens = int(model.pos_embed.shape[1])
    replaced = []

    for block_id, block in enumerate(model.blocks):
        if isinstance(block.attn, TimmAttentionWithDirectMLP):
            continue

        block.attn = TimmAttentionWithDirectMLP(
            original_attn=block.attn,
            num_tokens=num_tokens,
            row_hidden_dim=row_hidden_dim,
            element_hidden_dim=element_hidden_dim,
        )
        replaced.append(f"blocks.{block_id}.attn")

    return replaced


def unfreeze_all(model):
    for p in model.parameters():
        p.requires_grad = True


def parameter_counts(model):
    total = sum(p.numel() for p in model.parameters())
    trainable = sum(
        p.numel() for p in model.parameters()
        if p.requires_grad
    )
    return trainable, total


def print_trainable_parameters(model, title=""):
    trainable, total = parameter_counts(model)

    if title:
        print(title)

    print(
        f"Trainable: {trainable:,} / {total:,} "
        f"({100.0 * trainable / total:.4f}%)"
    )
```

## 6. Training / evaluation / logging / resume helpers

`last.pt` is written after every completed epoch. If the run is interrupted,
rerunning the notebook with `AUTO_RESUME=True` resumes from the next epoch.

Checkpoints are always deserialized to **CPU first** and then moved to the
currently selected CUDA device. Optimizer-state tensors are also explicitly
moved, so switching from one CUDA index to another is supported.

`training_log.txt` receives one line after every completed epoch.


```python
# ============================================================
# Training / evaluation / checkpoint / logging helpers
# ============================================================

criterion = nn.CrossEntropyLoss()


def autocast_context(device):
    if USE_AMP and device.type == "cuda":
        return torch.autocast(
            device_type="cuda",
            dtype=AMP_DTYPE,
        )
    return nullcontext()


def sync_device(device):
    if device.type == "cuda":
        torch.cuda.synchronize(device)


def format_seconds(seconds):
    seconds = float(seconds)

    if seconds < 60:
        return f"{seconds:.1f}s"

    minutes, sec = divmod(seconds, 60)

    if minutes < 60:
        return f"{int(minutes)}m {sec:.1f}s"

    hours, minutes = divmod(minutes, 60)
    return f"{int(hours)}h {int(minutes)}m {sec:.1f}s"


def append_log(path, message):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    timestamp = datetime.now().isoformat(timespec="seconds")

    with path.open("a", encoding="utf-8") as f:
        f.write(f"[{timestamp}] {message}\n")
        f.flush()


def portable_torch_load(path):
    """
    Always deserialize tensors onto CPU first.
    This removes dependence on the CUDA index used when saving.
    """
    path = Path(path)

    try:
        return torch.load(
            path,
            map_location="cpu",
            weights_only=False,
        )
    except TypeError:
        return torch.load(
            path,
            map_location="cpu",
        )


def move_nested_to_device(obj, device):
    if torch.is_tensor(obj):
        return obj.to(device)

    if isinstance(obj, dict):
        return {
            k: move_nested_to_device(v, device)
            for k, v in obj.items()
        }

    if isinstance(obj, list):
        return [
            move_nested_to_device(v, device)
            for v in obj
        ]

    if isinstance(obj, tuple):
        return tuple(
            move_nested_to_device(v, device)
            for v in obj
        )

    return obj


def optimizer_to(optimizer, device):
    for param_state in optimizer.state.values():
        for key, value in list(param_state.items()):
            param_state[key] = move_nested_to_device(
                value,
                device,
            )


def atomic_torch_save(obj, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    tmp_path = path.with_name(path.name + ".tmp")
    torch.save(obj, tmp_path)
    os.replace(tmp_path, path)


def expected_checkpoint_config(model):
    return {
        "model_name": MODEL_NAME,
        "num_classes": NUM_CLASSES,
        "num_tokens": int(model.pos_embed.shape[1]),
        "learned_norm_hidden": LEARNED_NORM_HIDDEN,
        "attention_row_hidden": ATTENTION_ROW_HIDDEN,
        "attention_element_hidden": ATTENTION_ELEMENT_HIDDEN,
        "attention_mapping": ATTENTION_MAPPING,
        "run_variant": RUN_VARIANT,
        "joint_epochs": JOINT_EPOCHS,
    }


def validate_checkpoint_config(ckpt, model):
    expected = expected_checkpoint_config(model)
    found = ckpt.get("config", None)

    if found is None:
        raise RuntimeError(
            "Checkpoint has no LENS config metadata. "
            "Automatic loading is refused because the variant "
            "cannot be verified."
        )

    mismatches = []

    for key, expected_value in expected.items():
        found_value = found.get(key, None)

        if found_value != expected_value:
            mismatches.append(
                f"{key}: checkpoint={found_value!r}, "
                f"current={expected_value!r}"
            )

    if mismatches:
        raise RuntimeError(
            "Checkpoint configuration mismatch:\n  "
            + "\n  ".join(mismatches)
            + "\nUse the matching notebook/output directory, "
              "or start a fresh run."
        )


@torch.no_grad()
def evaluate(model, loader, device):
    model.eval()

    total = 0
    correct1 = 0
    correct5 = 0
    loss_sum = 0.0

    sync_device(device)
    start_time = time.perf_counter()

    for images, labels in loader:
        images = images.to(
            device,
            non_blocking=True,
        )
        labels = labels.to(
            device,
            non_blocking=True,
        )

        with autocast_context(device):
            logits = model(images)
            loss = criterion(logits, labels)

        batch_size = labels.size(0)
        total += batch_size
        loss_sum += loss.item() * batch_size

        top5 = logits.topk(5, dim=1).indices
        correct1 += (
            top5[:, 0] == labels
        ).sum().item()
        correct5 += (
            top5.eq(labels[:, None])
            .any(dim=1)
            .sum()
            .item()
        )

    sync_device(device)
    elapsed = time.perf_counter() - start_time

    return {
        "loss": loss_sum / total,
        "top1": correct1 / total,
        "top5": correct5 / total,
        "time_sec": elapsed,
    }


def print_eval(
    model,
    loader,
    device,
    title,
    split_name="Validation",
):
    metrics = evaluate(
        model,
        loader,
        device,
    )

    print(f"\n[{title}]")
    print(
        f"{split_name}: loss={metrics['loss']:.4f}, "
        f"top1={metrics['top1']:.4f}, "
        f"top5={metrics['top5']:.4f}, "
        f"time={format_seconds(metrics['time_sec'])}"
    )

    return metrics


def load_model_checkpoint(
    model,
    path,
    device,
    validate_config=True,
    strict=True,
):
    path = Path(path)
    ckpt = portable_torch_load(path)

    if validate_config:
        validate_checkpoint_config(
            ckpt,
            model,
        )

    model.load_state_dict(
        ckpt["model"],
        strict=strict,
    )
    model.to(device)

    val = ckpt.get("val", {})

    print(
        f"Loaded checkpoint: {path} | "
        f"epoch={ckpt.get('epoch', '?')} | "
        f"val_top1={val.get('top1', '?')} | "
        f"current device={device}"
    )

    return ckpt


def train_joint_model(
    model,
    train_loader,
    val_loader,
    device,
    epochs,
    lr,
    weight_decay,
    out_dir,
    grad_clip_norm=1.0,
    auto_resume=True,
):
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    last_path = out_dir / "last.pt"
    best_path = out_dir / "best.pt"
    log_path = out_dir / "training_log.txt"
    history_path = out_dir / "history.json"

    trainable_params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    if not trainable_params:
        raise RuntimeError(
            "Joint training: no trainable parameters selected."
        )

    optimizer = torch.optim.AdamW(
        trainable_params,
        lr=lr,
        weight_decay=weight_decay,
    )

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=epochs,
    )

    trainable_count = sum(
        p.numel() for p in trainable_params
    )
    total_count = sum(
        p.numel() for p in model.parameters()
    )

    start_epoch = 1
    best_val_top1 = -float("inf")
    history = []

    if auto_resume and last_path.exists():
        ckpt = portable_torch_load(last_path)
        validate_checkpoint_config(
            ckpt,
            model,
        )

        model.load_state_dict(
            ckpt["model"],
            strict=True,
        )
        model.to(device)

        optimizer.load_state_dict(
            ckpt["optimizer"],
        )
        optimizer_to(
            optimizer,
            device,
        )

        scheduler.load_state_dict(
            ckpt["scheduler"],
        )

        start_epoch = int(ckpt["epoch"]) + 1
        best_val_top1 = float(
            ckpt.get(
                "best_val_top1",
                -float("inf"),
            )
        )
        history = list(
            ckpt.get("history", [])
        )

        resume_message = (
            f"RESUME from {last_path} | "
            f"last completed epoch={start_epoch - 1} | "
            f"next epoch={start_epoch} | "
            f"device={device}"
        )

        print("\n" + resume_message)
        append_log(
            log_path,
            resume_message,
        )

    else:
        with log_path.open(
            "w",
            encoding="utf-8",
        ) as f:
            f.write("")

        with history_path.open(
            "w",
            encoding="utf-8",
        ) as f:
            json.dump([], f)

        append_log(
            log_path,
            (
                f"NEW RUN | variant={RUN_VARIANT} | "
                f"device={device} | "
                f"lr={lr:.3e} | epochs={epochs}"
            ),
        )

    print(
        f"\nJoint full-model training: "
        f"trainable parameters {trainable_count:,} / {total_count:,} "
        f"({100.0 * trainable_count / total_count:.4f}%)"
    )
    print(
        f"Joint full-model training: "
        f"lr={optimizer.param_groups[0]['lr']:.3e}, "
        f"target epochs={epochs}, "
        f"start epoch={start_epoch}"
    )
    print(
        "Text log:",
        log_path.resolve(),
    )

    append_log(
        log_path,
        (
            f"Trainable={trainable_count:,}/{total_count:,} "
            f"({100.0 * trainable_count / total_count:.4f}%)"
        ),
    )

    if start_epoch > epochs:
        message = (
            f"Training already completed through epoch "
            f"{start_epoch - 1}; target is {epochs}. "
            f"No additional epochs are needed."
        )
        print(message)
        append_log(
            log_path,
            message,
        )
        return {
            "best_val_top1": best_val_top1,
            "history": history,
            "start_epoch": start_epoch,
        }

    for epoch in range(
        start_epoch,
        epochs + 1,
    ):
        model.train()

        total = 0
        correct1 = 0
        loss_sum = 0.0
        max_grad_norm = 0.0

        sync_device(device)
        train_start = time.perf_counter()

        for images, labels in train_loader:
            images = images.to(
                device,
                non_blocking=True,
            )
            labels = labels.to(
                device,
                non_blocking=True,
            )

            optimizer.zero_grad(
                set_to_none=True,
            )

            with autocast_context(device):
                logits = model(images)
                loss = criterion(
                    logits,
                    labels,
                )

            loss.backward()

            if grad_clip_norm is not None:
                grad_norm = torch.nn.utils.clip_grad_norm_(
                    trainable_params,
                    max_norm=grad_clip_norm,
                )

                try:
                    grad_norm_value = float(
                        grad_norm.detach().cpu()
                    )

                    if math.isfinite(grad_norm_value):
                        max_grad_norm = max(
                            max_grad_norm,
                            grad_norm_value,
                        )
                except Exception:
                    pass

            optimizer.step()

            batch_size = labels.size(0)
            total += batch_size
            loss_sum += (
                loss.item() * batch_size
            )
            correct1 += (
                logits.argmax(dim=1) == labels
            ).sum().item()

        sync_device(device)
        train_time = (
            time.perf_counter()
            - train_start
        )

        scheduler.step()

        train_loss = loss_sum / total
        train_top1 = correct1 / total

        val_metrics = evaluate(
            model,
            val_loader,
            device,
        )

        current_lr = (
            scheduler.get_last_lr()[0]
        )

        record = {
            "epoch": epoch,
            "lr": current_lr,
            "train_loss": train_loss,
            "train_top1": train_top1,
            "val_loss": val_metrics["loss"],
            "val_top1": val_metrics["top1"],
            "val_top5": val_metrics["top5"],
            "train_time_sec": train_time,
            "val_time_sec": val_metrics["time_sec"],
            "max_finite_grad_norm_before_clipping": max_grad_norm,
        }
        history.append(record)

        line = (
            f"Epoch {epoch:03d}/{epochs} | "
            f"lr={current_lr:.3e} | "
            f"train loss={train_loss:.4f} "
            f"top1={train_top1:.4f} | "
            f"val loss={val_metrics['loss']:.4f} "
            f"top1={val_metrics['top1']:.4f} "
            f"top5={val_metrics['top5']:.4f} | "
            f"train time={format_seconds(train_time)} | "
            f"val time={format_seconds(val_metrics['time_sec'])} | "
            f"max grad norm={max_grad_norm:.3e}"
        )

        print(line)
        append_log(
            log_path,
            line,
        )

        is_new_best = (
            val_metrics["top1"]
            > best_val_top1
        )

        if is_new_best:
            best_val_top1 = (
                val_metrics["top1"]
            )

        checkpoint_common = {
            "stage": "Joint full-model LENS",
            "epoch": epoch,
            "model": model.state_dict(),
            "val": val_metrics,
            "best_val_top1": best_val_top1,
            "model_name": MODEL_NAME,
            "config": expected_checkpoint_config(model),
            "history": history,
            "saved_from_device": str(device),
        }

        if is_new_best:
            atomic_torch_save(
                checkpoint_common,
                best_path,
            )

        atomic_torch_save(
            {
                **checkpoint_common,
                "optimizer": optimizer.state_dict(),
                "scheduler": scheduler.state_dict(),
            },
            last_path,
        )

        with history_path.open(
            "w",
            encoding="utf-8",
        ) as f:
            json.dump(
                history,
                f,
                indent=2,
            )

    done_message = (
        f"DONE | best validation top1={best_val_top1:.4f} | "
        f"last completed epoch={epochs}"
    )

    print(done_message)
    append_log(
        log_path,
        done_message,
    )

    return {
        "best_val_top1": best_val_top1,
        "history": history,
        "start_epoch": start_epoch,
    }
```

## 8. Direct joint full-LENS fine-tuning (pure-MLP attention)

LayerNorm and the analytical Softmax/exponential attention mapping are replaced
**simultaneously**, the whole converted DeiT-Tiny is unfrozen, and all parameters
are jointly fine-tuned with one learning rate.

If `last.pt` exists and `AUTO_RESUME=True`, model, optimizer, scheduler,
best validation score, and completed history are restored automatically.


```python
# ============================================================
# Install full LENS and jointly fine-tune the entire model
# ============================================================

replaced_norms = replace_all_layernorms(
    model,
    hidden_dim=LEARNED_NORM_HIDDEN,
)

replaced_attn = replace_all_attention_mapping(
    model,
    row_hidden_dim=ATTENTION_ROW_HIDDEN,
    element_hidden_dim=ATTENTION_ELEMENT_HIDDEN,
)

model = model.to(device)
unfreeze_all(model)

print(
    f"Replaced {len(replaced_norms)} LayerNorm modules."
)
for name in replaced_norms:
    print("  LN:", name)

print(
    f"Replaced analytical attention mapping in {len(replaced_attn)} attention modules."
)
for name in replaced_attn:
    print("  Attn:", name)

print("\n=== Joint full-LENS training ===")
print_trainable_parameters(model)

training_result = train_joint_model(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    device=device,
    epochs=JOINT_EPOCHS,
    lr=JOINT_LR,
    weight_decay=WEIGHT_DECAY,
    out_dir=JOINT_OUT_DIR,
    grad_clip_norm=GRAD_CLIP_NORM,
    auto_resume=AUTO_RESUME,
)

if not BEST_CKPT.exists():
    raise FileNotFoundError(
        f"Best checkpoint was not created: {BEST_CKPT}"
    )

load_model_checkpoint(
    model,
    BEST_CKPT,
    device,
    validate_config=True,
    strict=True,
)

final_metrics = print_eval(
    model,
    val_loader,
    device,
    title="Best full-LENS DeiT-Tiny",
    split_name="Validation",
)

print("\n=== Summary ===")
print(
    f"Standard pretrained top1: "
    f"{standard_metrics['top1']:.4f} | "
    f"top5: {standard_metrics['top5']:.4f}"
)
print(
    f"Full LENS top1:           "
    f"{final_metrics['top1']:.4f} | "
    f"top5: {final_metrics['top5']:.4f}"
)
print(
    f"Top-1 delta vs standard:  "
    f"{100.0 * (final_metrics['top1'] - standard_metrics['top1']):+.3f} pp"
)
print("Best checkpoint:", BEST_CKPT.resolve())
print("Resume checkpoint:", LAST_CKPT.resolve())
print("Epoch text log:", LOG_FILE.resolve())
```

## Notes

- Model: plain `deit_tiny_patch16_224.fb_in1k` (not distilled), 197 tokens.
- Direct pure-MLP attention uses `197 -> 32 -> 1` for the row state and a shared `2 -> 16 -> 1` element MLP.
- Hidden activations are ReLU; only the second MLP's final output uses Sigmoid.
- Attention coefficients are divided by `128`, so each row sum is strictly below `197/128 = 1.5390625`.
- There is no Softmax, `exp`, or `exp2` in the learned score-to-weight mapping.
- LearnedNorm `b`/beta is intentionally **unclamped**.
- `AUTO_RESUME=True` resumes after the most recently completed epoch. If the
  process stops inside an epoch, that epoch is repeated.
- All checkpoint loads go to CPU first, then to the selected CUDA device.
- This variant uses its own output directory and checkpoint metadata, so
  accidentally loading a Base-2/other LENS checkpoint is rejected.
- Watch progress from a shell with:
  `tail -f <run-dir>/training_log.txt`
- To deliberately start a fresh run, use a new `JOINT_OUT_DIR` or remove the
  previous run directory.
